## 1. A table is rows and columns

A database holds tables. A table is a grid: each column has a name and a type,
each row is one thing (one game, one user, one earthquake).

| Type | Holds |
|---|---|
| `INTEGER` | whole numbers |
| `REAL` | decimals |
| `TEXT` | words, dates |

A column can also have rules: `PRIMARY KEY` (unique id), `NOT NULL` (must have a value), `UNIQUE` (no repeats).

## 2. CREATE TABLE

We'll practise with a table of video games.

```sql
CREATE TABLE games (
    id     INTEGER PRIMARY KEY AUTOINCREMENT,
    title  TEXT NOT NULL,
    genre  TEXT NOT NULL,
    year   INTEGER,
    rating REAL
);
```

`AUTOINCREMENT` means the database numbers the rows for you. Run it once only: running it again says the table already exists.

## 3. INSERT: add rows

```sql
INSERT INTO games (title, genre, year, rating)
VALUES ('Hollow Knight', 'Platformer', 2017, 9.1);
```

Several rows in one go:

```sql
INSERT INTO games (title, genre, year, rating) VALUES
    ('Celeste',        'Platformer', 2018, 8.9),
    ('Stardew Valley', 'Simulation', 2016, 8.7),
    ('Portal 2',       'Puzzle',     2011, 9.5),
    ('Tetris',         'Puzzle',     1984, 9.0),
    ('Hades',          'Roguelike',  2020, 9.3),
    ('Minecraft',      'Sandbox',    2011, 8.8);
```

Text goes in single quotes. Numbers do not. You did not give an `id`: the database did.

## 4. SELECT: ask for rows

Everything:

```sql
SELECT * FROM games;
```

`*` means "all columns". Just the ones you want:

```sql
SELECT title, rating FROM games;
```

You should see 7 rows.

## 5. WHERE: only the rows that match

```sql
SELECT title, year FROM games WHERE genre = 'Puzzle';
```

Comparisons: `=`, `!=`, `>`, `<`, `>=`, `<=`. Combine with `AND` / `OR`:

```sql
SELECT title, year, rating FROM games WHERE year > 2015 AND rating >= 9.0;
```

Search for part of a word with `LIKE` (`%` means "anything"):

```sql
SELECT title FROM games WHERE title LIKE '%craft%';
```

## 6. ORDER BY and LIMIT

```sql
SELECT title, rating FROM games ORDER BY rating DESC;
```

`DESC` is biggest first; `ASC` is smallest first. Keep only the top three:

```sql
SELECT title, rating FROM games ORDER BY rating DESC LIMIT 3;
```

## 7. Count and summarise

```sql
SELECT COUNT(*) FROM games;
```

`AVG`, `MAX`, `MIN`, `SUM` work on a number column:

```sql
SELECT AVG(rating), MAX(rating), MIN(year) FROM games;
```

`GROUP BY` does the sum for each group. This one answers "how many games per genre, and what is the average rating?":

```sql
SELECT genre, COUNT(*) AS how_many, AVG(rating) AS average_rating
FROM games
GROUP BY genre
ORDER BY how_many DESC;
```

`AS` gives a result column a clearer name. This kind of query will power your dashboard in Week 11.

## 8. UPDATE and DELETE

Always use `WHERE`. Without it, the change hits every row.

```sql
UPDATE games SET rating = 9.2 WHERE title = 'Tetris';
```

Check it:

```sql
SELECT title, rating FROM games WHERE title = 'Tetris';
```

Delete:

```sql
DELETE FROM games WHERE year < 1990;
```

> A safe habit. Before an `UPDATE` or `DELETE`, write the same `WHERE` in a `SELECT` first and look at which rows it finds.
> If those are the rows you mean, change `SELECT *` to `DELETE`.

```sql
SELECT COUNT(*) FROM games;
```

You should be down to 6 rows now.

## Exercises

Write one query for each. Try before you open the answer.

3. How many games are there in each genre?
4. What is the average rating of the `Platformer` games?
5. Change the rating of `Minecraft` to `9.0`.
6. Delete every game with a rating below 8.8. (Run a SELECT with the same condition first.)

<details><summary>Answers</summary>

```sql
SELECT * FROM games WHERE genre = 'Puzzle';

SELECT title, rating FROM games WHERE year > 2015 ORDER BY rating DESC;

SELECT genre, COUNT(*) FROM games GROUP BY genre;

SELECT AVG(rating) FROM games WHERE genre = 'Platformer';

UPDATE games SET rating = 9.0 WHERE title = 'Minecraft';

SELECT * FROM games WHERE rating < 8.8;   -- check first
DELETE FROM games WHERE rating < 8.8;
```
